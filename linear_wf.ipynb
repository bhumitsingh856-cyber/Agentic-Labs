```python
from langgraph.graph import StateGraph , START , END
from typing import TypedDict
```

```python
class ADD(TypedDict):
    num1:int
    num2:int
    result:int
```

```python
def add(state:ADD) -> ADD:
    num1=state['num1']
    num2=state['num2']
    state['result']=num1+num2
    return state
```

```python
graph = StateGraph(ADD)

graph.add_node("Addition",add)

graph.add_edge(START,'Addition')
graph.add_edge('Addition',END)

workflow=graph.compile()
```

```python
a=workflow.invoke({"num1":1,"num2":1})
print(a['result'])
```

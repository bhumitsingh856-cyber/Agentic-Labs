```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class State(TypedDict):
    number:int
```

```python
def reduct(state:State):
    state['number']-=1
    return state
def condition(state:State):
    if(state['number']==1):
        return "end"
    else :
        return "repeat"
```

```python
graph=StateGraph(State)
graph.add_node("reduct",reduct)

graph.add_edge(START,"reduct")
graph.add_conditional_edges("reduct",condition,{"end":END,"repeat":"reduct"})

wf=graph.compile()
```

```python
print(wf.invoke({"number":5}))
```

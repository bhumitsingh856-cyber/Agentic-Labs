```python
from langgraph.store.memory import InMemoryStore 
from llm import embedding_model 
```

```python
store=InMemoryStore(index={"embed":embedding_model})
```

```python
namespace=("user","u1")
# namespace , key , value
# Add data to store
store.put(namespace, "1", {
    "data": "User is learning LangChain in Python"
})

store.put(namespace, "2", {
    "data": "User already knows LangChain in JavaScript"
})

store.put(namespace, "3", {
    "data": "User is learning LangGraph for AI workflows"
})

store.put(namespace, "4", {
    "data": "User uses Next.js for frontend development"
})

store.put(namespace, "5", {
    "data": "User prefers Groq models for inference"
})

store.put(namespace, "6", {
    "data": "User is building AI-powered applications"
})

store.put(namespace, "7", {
    "data": "User is interested in FastAPI for AI backends"
})

store.put(namespace, "8", {
    "data": "User is learning embeddings and RAG systems"
})
```

```python
namespace2=("user","u2")
store.put(namespace2,"2",{"data":"Agents can use tools"})
```

```python
# Retrieve specif key data from store
store.get(namespace,"1")
store.get(namespace,"2")
store.get(namespace2,"2")
```

```python
# Get all data in a namespace
for i in store.search(namespace):
    print(i.value['data'])
```

```python
# search for relavant memory
store.search(namespace,query="preferances",limit=2)
```

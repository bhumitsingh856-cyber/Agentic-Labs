```python
from langgraph.graph import StateGraph ,START,END
from typing import TypedDict
```

```python
class State(TypedDict):
    a:int
    b:int
    max_iteration:int
```

```python
def sq(state:State):
    state['b']=state['a']*2
    return state
def condition(state:State) : 
    if(state['b']==0 or state['max_iteration']>5):
        return 'zero'
    else : return "one"
def one(state:State):
    state['b']-=1
    state['max_iteration']+=1
    print(state)
    return state
```

```python
graph=StateGraph(State)
graph.add_node("sq",sq)
graph.add_node("one",one)

graph.add_edge(START,"sq")
graph.add_conditional_edges("sq",condition,{"zero":END,"one":"one"})
graph.add_edge("one","sq")


workflow=graph.compile()
```

```python
a=workflow.invoke({"a":3,"b":0,"max_iteration":0})
print(a)
```

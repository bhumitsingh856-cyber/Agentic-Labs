```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from llm import llm
```

```python
class SaState(TypedDict):
    topic:str
    essay:str
    language_score:float
    quality_score:float
    knowledge_score:float
  
```

```python
def essay(state:SaState):
    topic=state["topic"]
    prompt=f"You are a essay writer , write in a childish tone essay on topic - {topic}"
    res=llm.invoke(prompt).content
    state['essay']=res
    return state
```

```python
def langrate(state:SaState):
    topic=state['topic']
    essay=state['essay']
    prompt=f"You are a essay analyzer , your work is to rate the essay from 0-10 in terms of language, return only the score , topic - {topic} essay - {essay}"
    res=llm.invoke(prompt).content
    return {"language_score":res}
```

```python
def qualrate(state:SaState):
    topic=state['topic']
    essay=state['essay']
    prompt=f"You are a essay analyzer , your work is to rate the essay from 0-10 in terms of quality, return only the score , topic - {topic} essay - {essay}"
    res=llm.invoke(prompt).content
    return {"quality_score":res}
```

```python
def knowrate(state:SaState):
    topic=state['topic']
    essay=state['essay']
    prompt=f"You are a essay analyzer , your work is to rate the essay from 0-10 in terms of knowledge, return only the score , topic - {topic} essay - {essay}"
    res=llm.invoke(prompt).content
    return {"knowledge_score":res}
```

```python
graph=StateGraph(SaState)

graph.add_node("writeEssay",essay)
graph.add_node("langrate",langrate)
graph.add_node("qualrate",qualrate)
graph.add_node("knowrate",knowrate)
graph.add_node("score",score)

graph.add_edge(START,"writeEssay")
graph.add_edge("writeEssay","langrate")
graph.add_edge("writeEssay","qualrate")
graph.add_edge("writeEssay","knowrate")
graph.add_edge("langrate",END)
graph.add_edge("qualrate",END)
graph.add_edge("knowrate",END)
 

workflow=graph.compile()
```

```python
res=workflow.invoke({"topic":"LLM"})
print(res)
```
